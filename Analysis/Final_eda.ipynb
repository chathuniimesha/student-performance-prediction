```python
import pandas as pd
```

```python
df = pd.read_csv(r"D:\student-performance-prediction\Data\cleaned_student_data.csv")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.histplot(df['GPA'],
             bins=10,
             kde=True,
             color='skyblue',
            )
plt.title('GPA Distribution')
plt.xlabel('frequancy')
plt.ylabel('GPA')
plt.show()
```

```python
fig, ax=plt.subplots(figsize=(8,5))
sns.scatterplot(x='GPA',
                y='StudyTimeWeekly',
                data=df,
                ax=ax,
                color='purple',
                alpha=0.7,
                hue='GPA',
                palette='viridis')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(
y='ParentalSupport',
x='GPA',
data=df,
palette='viridis',
hue='ParentalEducation',
orient='h'
)
plt.xlabel('Average GPA')
plt.ylabel('Parental support')
plt.title('Average GPA by parent support and education')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(
    y='GPA',
    x='GradeClass',
    data=df,
    color= 'blue'



)
plt.title('GPA by Grade class ')
plt.show()
```
